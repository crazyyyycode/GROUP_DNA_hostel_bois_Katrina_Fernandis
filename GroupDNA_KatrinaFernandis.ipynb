```python
#AI was used casue i was not able to figure out the errors
def parse_chat_file(file_path):
    messages = []
    system_messages = 0
    media_messages = 0
    deleted_messages = 0
    skipped_lines = 0

    with open(file_path, "r", encoding="utf-8") as f:
        lines = f.readlines()

    current_sender = None
    current_time = None
    current_msg = ""

    # Regex-like parsing using string methods
    # Format: DD/MM/YY, HH:MM:SS - Sender: Message
    for line in lines:
        line = line.strip()
        if not line:
            skipped_lines += 1
            continue

        # Check for timestamp pattern
        try:
            # Find the first dash to separate timestamp from rest
            if " - " in line:
                timestamp_part = line.split(" - ")[0]
                # Validate date format
                datetime.strptime(timestamp_part, "%d/%m/%y, %H:%M:%S")

                # If we had a previous message, save it
                if current_sender and current_msg:
                    messages.append({
                        "time": current_time,
                        "sender": current_sender,
                        "text": current_msg
                    })

                # Parse new line
                rest = line.split(" - ", 1)[1]
                if ":" in rest:
                    current_sender = rest.split(": ")[0]
                    current_msg = rest.split(": ", 1)[1] if ": " in rest else ""
                    current_time = datetime.strptime(timestamp_part, "%d/%m/%y, %H:%M:%S")
                else:
                    # System message or malformed
                    system_messages += 1
                    skipped_lines += 1
                    current_sender = None
                    current_msg = ""
            else:
                # Continuation of previous message
                if current_sender:
                    current_msg += " " + line
                else:
                    skipped_lines += 1

        except ValueError:
            # Not a valid timestamp line
            if current_sender:
                current_msg += " " + line
            else:
                skipped_lines += 1

    # Save last message
    if current_sender and current_msg:
        messages.append({
            "time": current_time,
            "sender": current_sender,
            "text": current_msg
        })

    print(f"Parsed {len(messages)} messages.")
    print(f"Skipped {skipped_lines} lines.")
    return messages

messages = parse_chat_file("hostel_bois.txt")
```
